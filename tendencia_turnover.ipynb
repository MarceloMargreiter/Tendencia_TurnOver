```python
X_cat = employee_df[['BusinessTravel', 'Department', 'EducationField', 'Gender', 'JobRole', 'MaritalStatus']]
X_cat
```

```python
# Dummy Variables (variáveis Dummy) são variáveis binárias (0 ou 1) que representam categorias. 
# Elas são usadas para converter variáveis categóricas em um formato numérico que pode ser usado em modelos de machine learning. pois se usarmos variáveis categóricas diretamente, o modelo pode interpretar os valores como ordinais, o que não é o caso. 
# Por exemplo, se tivermos uma variável "Cor" com os valores "Vermelho", "Verde" e "Azul", o modelo pode interpretar "Verde" como maior que "Vermelho" e menor que "Azul", o que não faz sentido. 
# Ao criar variáveis dummy, cada categoria é representada por uma coluna separada, com valores 0 ou 1 indicando a presença ou ausência da categoria.

#Exemplos
# Male    Female
# 1       0
# 0       1

# Single    Married    Divorced
# 1         0          0            
# 0         1          0
# 0         0          1
from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder()
X_cat = onehotencoder.fit_transform(X_cat).toarray()
```
