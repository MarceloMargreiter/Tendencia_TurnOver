# Visualisação/Exploração de dados

```python
# Função ambda usada para transformar os valores dos atributos OverTime, Over18 e Attrition em valores binários (0 e 1) para facilitar a análise.
employee_df['Attrition'] = employee_df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
employee_df['OverTTime'] = employee_df['OverTime'].apply(lambda x: 1 if x == 'Yes' else 0)
employee_df['Over18'] = employee_df['Over18'].apply(lambda x: 1 if x == 'Yes' else 0)
```
